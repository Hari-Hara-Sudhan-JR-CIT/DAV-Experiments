```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm

# Load Dataset
uci_diabetes = pd.read_csv("uci_diabetes.csv")

# Create Figure
plt.figure(figsize=(12, 5))

# -----------------------------
# Normal Curve for Glucose
# -----------------------------
plt.subplot(1, 2, 1)

sns.histplot(
    uci_diabetes["Glucose"],
    bins=20,
    kde=True,
    stat="density",
    color="skyblue"
)

x = np.linspace(
    uci_diabetes["Glucose"].min(),
    uci_diabetes["Glucose"].max(),
    100
)

plt.plot(
    x,
    norm.pdf(
        x,
        uci_diabetes["Glucose"].mean(),
        uci_diabetes["Glucose"].std()
    ),
    color="red",
    linewidth=2,
    label="Normal Distribution"
)

plt.title("Normal Curve - Glucose")
plt.xlabel("Glucose")
plt.ylabel("Density")
plt.legend()

# -----------------------------
# Normal Curve for BMI
# -----------------------------
plt.subplot(1, 2, 2)

sns.histplot(
    uci_diabetes["BMI"],
    bins=20,
    kde=True,
    stat="density",
    color="lightgreen"
)

x = np.linspace(
    uci_diabetes["BMI"].min(),
    uci_diabetes["BMI"].max(),
    100
)

plt.plot(
    x,
    norm.pdf(
        x,
        uci_diabetes["BMI"].mean(),
        uci_diabetes["BMI"].std()
    ),
    color="red",
    linewidth=2,
    label="Normal Distribution"
)

plt.title("Normal Curve - BMI")
plt.xlabel("BMI")
plt.ylabel("Density")
plt.legend()

plt.tight_layout()
plt.show()
```
